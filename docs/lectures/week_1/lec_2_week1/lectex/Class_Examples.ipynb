```python
measurements = [1.0, 2.0]
alias = measurements
alias.append(3.0)
# measurements is now [1.0, 2.0, 3.0]
```

```python
print(f"alias: {alias}, measurements: {measurements}") # measurements is now [1.0, 2.0, 3.0] why? because alias and measurements refer to the same list object in memory.
```

```python
independent_copy = measurements.copy() # independent copy
```

```python
print(f"measurements: {measurements}, alias: {alias}, independent_copy: {independent_copy}") # independent_copy is now [1.0, 2.0, 3.0] but it is a separate list object in memory.
```

```python
# Create two identical lists
a = [1, 2, 3]
b = [1, 2, 3]

# Check their values
print(a == b)  # True (Same contents)

# Check their memory addresses
print(a is b)  # False (Different memory)

# Make 'c' point to 'a'
c = a
print(a is c)  # True (Same exact object)
```

```python
drone_position = [2.0, 1.0, -4.0]
drone_battery = 75.0

def calculate_altitude(position):
    return -position[2]

def battery_is_low(battery):
    return battery < 25.0
```

```python
class DroneState:
    def __init__(self, position, battery):
        self.position = position
        self.battery = battery

    def calculate_altitude(self):
        return -self.position[2]

    def battery_is_low(self):
        return self.battery < 25.0



# Create an instance of the DroneState class; the constructor (__init__) is called automatically when the object is created, and it initializes the position and battery attributes of the drone.
drone = DroneState(
    position=[2.0, 1.0, -4.0],
    battery=75.0
)

#drone_with_no_initaition = DroneState() # This will raise a TypeError because the constructor requires two arguments (position and battery) but none were provided. 


altitude = drone.calculate_altitude()
low_battery = drone.battery_is_low()
```

```python
print(f"Drone altitude: {altitude} meters") # This will print 4.0, which is the altitude of the drone in meters.
```

```python
class DroneState:
    def __init__(self, north_m, east_m, down_m):
        self.north_m = north_m
        self.east_m = east_m
        self.down_m = down_m

    def altitude_m(self):
        return -self.down_m

state = DroneState(2.0, -1.0, -4.0)
print(state.altitude_m())
```
